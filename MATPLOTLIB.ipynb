```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [80, 70, 90, 60]

plt.bar(students, marks)
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

students = ["A", "B", "C", "D"]
marks = [80, 70, 90, 60]

plt.barh(students, marks)
plt.title("Student Marks")
plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ["Python", "Java", "C++", "C"]
values = [40, 30, 20, 10]

plt.pie(values, labels=labels, autopct="%1.1f%%")
plt.title("Programming Languages")
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 30, 40, 50]
y2 = [5, 15, 25, 35, 45]

plt.plot(x, y1, label="Line 1")
plt.plot(x, y2, label="Line 2")

plt.title("Multiple Line Plot")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 25, 15, 30]

plt.plot(x, y, marker="o", linestyle="--")
plt.title("Line with Markers")
plt.grid(True)
plt.show()
```

```python
import matplotlib.pyplot as plt

marks = [40, 50, 55, 60, 65, 70, 75, 90, 100]

plt.boxplot(marks)
plt.title("Box Plot")
plt.ylabel("Marks")
plt.show()
```
